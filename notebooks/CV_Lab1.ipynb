```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

print("Please upload a sample image (e.g., sample.jpg):")
uploaded = files.upload()

filename = next(iter(uploaded))
print(f"Successfully uploaded: {filename}")
```

```python
# 1. Load the color image using OpenCV
img = cv2.imread(filename)

# 2. Display the original image
print("Original Image:")
cv2_imshow(img)

# 3. Print the image shape and identify dimensions
print("\nImage Properties:")
print(f"Shape: {img.shape}")
print(f"Height: {img.shape[0]} pixels")
print(f"Width: {img.shape[1]} pixels")
print(f"Channels: {img.shape[2]} (BGR Color Ordering)")

# 4. Convert the image to grayscale and display it
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("\nGrayscale Image:")
cv2_imshow(gray_img)

# 5. Examine pixel values at specific coordinates (e.g., y=100, x=100)
pixel_value = img[100, 100]
print(f"\nOriginal pixel value at (100, 100): {pixel_value} (B, G, R)")

# 6. Change a pixel's color to a new value (Red = 0, 0, 255 in BGR) and observe
img_modified = img.copy()
img_modified[100, 100] = [0, 0, 255]
print("Pixel color changed to Red at (100, 100).")
```

```python
# 1. Crop a region of interest (ROI) from the image
# Format is img[y_start:y_end, x_start:x_end]
roi_cropped = img[50:250, 50:250]
print("Cropped Image (ROI):")
cv2_imshow(roi_cropped)

# 2. Resize the image to a smaller/larger dimension (e.g., 300x300 pixels)
resized_img = cv2.resize(img, (300, 300))
print("\nResized Image (300x300):")
cv2_imshow(resized_img)

# 3. Rotate the image by 90 degrees
rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
print("\nImage Rotated by 90 Degrees:")
cv2_imshow(rotated_img)
```
